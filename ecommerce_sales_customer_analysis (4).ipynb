```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

df = pd.read_csv("sales and customer analytics.csv")
```

```python
print(df.columns.tolist())
```

```python
category_sales = df.groupby('category')['total_amount'].sum().sort_values(ascending=False)

print(category_sales)
```

```python
import matplotlib.pyplot as plt

category_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby('region')['total_amount'].sum().sort_values(ascending=False)

print(region_sales)
```

```python
region_sales.plot(kind='bar', figsize=(8, 5))

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['order_date'] = pd.to_datetime(df['order_date'])

daily_sales = df.groupby('order_date')['total_amount'].sum()

print(daily_sales.head())
```

```python
plt.figure(figsize=(10, 5))

plt.plot(daily_sales.index, daily_sales.values)

plt.title('Sales Trend Over Time')
plt.xlabel('Order Date')
plt.ylabel('Total Sales Amount')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
payment_counts = df['payment_method'].value_counts()

print(payment_counts)
```

```python
payment_counts.plot(kind='bar', figsize=(8, 5))

plt.title('Number of Orders by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['customer_age'], bins=20, edgecolor='black')

plt.title('Customer Age Distribution')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

print(correlation)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(correlation, annot=True, fmt='.2f')

plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
returned_counts = df['returned'].value_counts()

print(returned_counts)
```

```python
returned_counts.plot(kind='bar', figsize=(7, 5))

plt.title('Returned vs Non-Returned Orders')
plt.xlabel('Returned Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit = df.groupby('category')['profit_margin'].mean().sort_values(ascending=False)

print(category_profit)
```

```python
category_profit.plot(kind='bar', figsize=(8, 5))

plt.title('Average Profit Margin by Category')
plt.xlabel('Category')
plt.ylabel('Average Profit Margin')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_delivery = df.groupby('region')['delivery_time_days'].mean().sort_values()

print(region_delivery)
```

```python
region_delivery.plot(kind='bar', figsize=(8, 5))

plt.title('Average Delivery Time by Region')
plt.xlabel('Region')
plt.ylabel('Average Delivery Time (Days)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
